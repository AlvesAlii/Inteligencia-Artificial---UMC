10.Sequência de Fibonacci

Imprima os primeiros 10 números da sequência de Fibonacci.

```python
# Define a quantidade inicial de termos desejada para a sequência de Fibonacci
DtTermo = 10
# Ajusta o valor limite somando 1 para adequar a parada do laço
DtTermo = DtTermo + 1

# Inicializa os dois primeiros números da sequência
termo = 0
termo2 = 1

# Define o contador inicial de iterações
i = 1

# Executa o bloco enquanto o contador for menor que o limite definido
while i < DtTermo:
    # Exibe o termo atual da sequência
    print(termo)
    # recebe termo2, e termo2 recebe a soma deles enquantos os atualiza simultaneamente
    termo, termo2 = termo2, termo + termo2
    # Incrementa o contador em 1 a cada volta
    i += 1
```

17. Dicionário de Contagem de Palavras

Conte a frequência de cada palavra em uma string fornecida pelo
usuário.

```python
# Define a frase em texto que será analisada
txt = "Aprender IA com o Fabiano é muito legal e divertido"

# Padroniza todas as letras em minúsculas e divide a frase em uma lista de palavras
palavras = txt.lower().split()

# Cria um dicionário vazio para armazenar a contagem de cada palavra
frequencia = {}

# Percorre individualmente cada palavra presente na lista
for palavra in palavras:
    # adiciona o contador da palavra no dicionário de forma segura
    frequencia[palavra] = frequencia.get(palavra, 0) + 1

# Exibe o dicionário resultante contendo as palavras e suas respectivas contagens
print(frequencia)
```

Escolhemos utilizar o método .lower() combinado com .split() por padronizar o texto em letras minúsculas e dividi-lo em uma lista de termos individuais, além disso, utilizamos a estrutura de dicionário {} para mapear a relação de chave e valor da palavra e contagem, ja o laço for utilizamos para percorrer a lista de palavras de forma limpa, utilizamos tambem o metodo .get(palavra, 0) + 1, por retornar a contagem atual da palavra ou o valor padrao 0 caso seja a primeira vez que ela apareça, evitando assim de acontecer erros de chave inexistentes no dicionario.

18. Número Primo

Crie um programa que verifique se um número fornecido é primo.


```python
# Solicita um número ao usuário
num = int(input("Digite o Número que deseja saber se é primo ou não: "))

# Filtra os números menores ou iguais a 1 não considerados primos
if num <= 1:
    primo = False
else:
    # Define a variável booleana como verdadeira para testar os divisores
    primo = True

    # Verifica os possíveis divisores do 2 até a raiz quadrada do número
    for i in range(2, int(num ** 0.5) + 1):
        # Se for divisível por qualquer número desse intervalo, não é primo
        if num % i == 0:
            primo = False
            break

# Exibe a mensagem final de acordo com a validação da variável primo
if primo:
    print(f"O número {num} é primo")
else:
    print(f"O número {num} não é primo")
```

22. Criptografia de Cifra de César
o Crie uma função para criptografar uma mensagem usando a cifra de
César.


```python
# Solicita a mensagem original ao usuário
msgO = input("Digite a Mensagem que deseja criptografar: ")

# Solicita a chave de deslocamento e a converte para número inteiro
chave = int(input("Digite o valor da chave para o deslocamento das letras: "))


# Define a função responsável por realizar a Cifra de César
def Ccesar(texto, deslocamento):
    resul = ""

    # Percorre individualmente cada caractere do texto de entrada
    for caractere in texto:
        # Verifica se o caractere atual é uma letra maiúscula
        if caractere.isupper():
            base = ord("A")
            # Calcula a nova posição circular da letra maiúscula e converte de volta para caractere
            Nletra = chr((ord(caractere) - base + deslocamento) % 26 + base)
            resul += Nletra

        # Verifica se o caractere atual é uma letra minúscula
        elif caractere.islower():
            base = ord("a")
            # Calcula a nova posição circular da letra minúscula e converte de volta para caractere
            Nletra = chr((ord(caractere) - base + deslocamento) % 26 + base)
            resul += Nletra

        # Mantém espaços, números e símbolos sem alteração
        else:
            resul += caractere

    # Retorna a mensagem completa criptografada
    return resul


# Executa a função passando o texto e a chave informados
cripMsg = Ccesar(msgO, chave)

# Exibe a mensagem criptografada
print(f"A mensagem ficou assim: {cripMsg}")
```

23. Descriptografar Cifra de César
o Crie uma função para descriptografar uma mensagem cifrada com a
cifra de César.

```python
# Solicita a mensagem que será descriptografada
msgO = input("Digite a Mensagem que deseja criptografar: ")

# Solicita a chave de deslocamento e a converte para número inteiro
chave = int(input("Digite o valor da chave para o deslocamento das letras: "))


# Define a função responsável por aplicar o deslocamento reverso na Cifra de César
def Ccesar(texto, deslocamento):
    resul = ""

    # Percorre individualmente cada caractere do texto de entrada
    for caractere in texto:
        # Verifica se o caractere atual é uma letra minúscula
        if caractere.isupper():
            base = ord("A")
            # Aplica o deslocamento para trás (subtração) mantendo a rotação dentro do alfabeto
            Nletra = chr((ord(caractere) - base - deslocamento) % 26 + base)
            resul += Nletra

        # Verifica se o caractere atual é uma letra minúscula
        elif caractere.islower():
            base = ord("a")
            # Aplica o deslocamento para trás (subtração) mantendo a rotação dentro do alfabeto
            Nletra = chr((ord(caractere) - base - deslocamento) % 26 + base)
            resul += Nletra

        # Mantém espaços, números e símbolos sem alteração
        else:
            resul += caractere

    # Retorna a mensagem completa criptografada
    return resul


# Executa a função passando o texto e a chave informados
cripMsg = Ccesar(msgO, chave)

# Exibe a mensagem descriptografada
print(f"A mensagem ficou assim: {cripMsg}")
```

24.Validação de CPF
o Implemente uma função para validar um CPF brasileiro.

```python
def validar_cpf(cpf):
    # 1. Limpa o texto mantendo apenas os números
    numeros = []
    for caractere in cpf:
        if caractere.isdigit():
            numeros.append(int(caractere))

    # 2. Valida se possui 11 dígitos
    if len(numeros) != 11:
        return False

    # 3. Evita CPFs com todos os números iguais (ex: 111.111.111-11)
    if numeros == [numeros[0]] * 11:
        return False

    # 4. Cálculo do 1º dígito verificador (pesos de 10 a 2)
    soma = 0
    peso = 10
    for i in range(9):
        soma += numeros[i] * peso
        peso -= 1

    resto = (soma * 10) % 11
    digito1 = 0 if resto == 10 else resto

    # 5. Cálculo do 2º dígito verificador (pesos de 11 a 2)
    soma = 0
    peso = 11
    for i in range(10):
        soma += numeros[i] * peso
        peso -= 1

    resto = (soma * 10) % 11
    digito2 = 0 if resto == 10 else resto

    # 6. Verifica se os dígitos calculados batem com os do CPF
    if numeros[9] == digito1 and numeros[10] == digito2:
        return True
    else:
        return False


# Exemplo de teste
#cpf_teste = "534.811.958-00"
cpf = input("Digite o cpf que quer verificar: ")

if validar_cpf(cpf):
    print("CPF VÁLIDO")
else:
    print("CPF INVÁLIDO")
```

29. Números FizzBuzz

```python
# Inicializa a variável de controle do laço com o valor inicial 1
i = 1

# Executa a repetição contínua enquanto a variável for menor ou igual a 100
while i <= 100:
    # Verifica se o número é divisível simultaneamente por 3 e por 5
    if i % 3 == 0 and i % 5 == 0:
        print("FizzBuzz")
    # Verifica se o número é divisível apenas por 3
    elif i % 3 == 0:
        print("Fizz")
    # Verifica se o número é divisível apenas por 5
    elif i % 5 == 0:
        print("Buzz")
    # Exibe o próprio número caso não seja múltiplo nem de 3 nem de 5
    else:
        print(i)

    # adiciona o valor de i em 1 unidade a cada iteração
    i += 1
```

30.Jogo da Forca
o Desenvolva um jogo simples da forca onde o usuário deve adivinhar
uma palavra.


```python
# Inicializa a palavra secreta, o conjunto de letras tentadas e a contagem de vidas
p, c, v = "python", set(), 6

# Executa o jogo enquanto houver vidas e restarem letras não descobertas
while v and set(p) - c:
    # Renderiza a palavra mascarando com '_' as letras ainda não adivinhadas e mostra as vidas
    print("".join(x if x in c else "_" for x in p), f"Vidas: {v}")

    # Captura a entrada com o operador walrus, reduz a vida se errou e registra a letra no conjunto
    v -= (t := input("Letra: ")) not in p
    c.add(t)

# Exibe a mensagem final de vitória ou de derrota informando a palavra oculta
print("Ganhou!" if v else f"Perdeu: {p}")
```

31.Ordenação por Bolha
o Implemente o algoritmo de ordenação por bolha (Bubble Sort).

```python
# Define a função BS (Bubble Sort) que recebe uma lista como parâmetro
def BS(lista):

  # len obtem o tamanho total da lista
  n = len(lista)

  # Loop externo: passa por todos os elementos da lista
  for i in range(n):
    # Inicializa a flag como False para verificar se houve troca nesta passagem
    trocado = False

    # realiza um loop interno que compara os elementos próximos
    # O limite 'n-i-1' evita comparar os elementos que já estão ordenados
    for j in range(0, n-i-1):
      # Se o elemento atual for maior que o próximo, eles estão fora de ordem
      if lista[j] > lista[j+1]:
        # Inverte a posição dos dois elementos
        lista[j], lista[j+1] = lista[j+1], lista[j]
      trocado = True
      if not trocado:
        break

  # Retorna a lista ordenada
  return lista

numeros = [64, 34, 25, 12, 22, 11, 90]
# Exibe a lista antes da ordenação
print(f"lista original: {numeros}")

# Chama a função BS passando a lista de números e imprime o resultado
lista_ordenada = BS(numeros)
print(f"lista ordenada: {lista_ordenada}")
```

32.Jogo da Velha
o Crie um jogo da velha para dois jogadores.


```python
import os

#função para criar o tabuleiro
def ver_tabuleiro(tabuleiro):
  #limpa o terminal para o jogo atualizar
  os.system('cls') if os.name == 'nt' else os.system('clear')

  print(f" {tabuleiro[0]} | {tabuleiro[1]} | {tabuleiro[2]} ")
  print("---|---|---")
  print(f" {tabuleiro[3]} | {tabuleiro[4]} | {tabuleiro[5]} ")
  print("---|---|---")
  print(f" {tabuleiro[6]} | {tabuleiro[7]} | {tabuleiro[8]} \n")

  #função para contabilizar vitoria
def Vitoria(tab, jogador):
  #condições de vitoria possiveis
  condicoes = [[0, 1, 2], [3, 4, 5], [6, 7, 8], [0, 3, 6], [1, 4, 7], [2, 5, 8], [0, 4, 8], [2, 4, 6]]
  return any(all(tab[pos] == jogador for pos in linha) for linha in condicoes)

def jogar():
  #inicializa o tabuleiro com numeros de 1 a 9 ao envez de 0 a 8
  tabuleiro = [str(i) for i in range(1, 10)]
  jogador_atual = "X"
  jogadas = 0

  #loop para as jogadas
  while jogadas < 9:
    ver_tabuleiro(tabuleiro)
    print(f"Turno do Jogador {jogador_atual}")

    try:
      escolha = int(input("Escolha uma posição (1-9): ")) - 1
    #valida se a jogada esta livre
      if escolha < 0 or escolha > 8  or tabuleiro[escolha] in ["X", "O"]:
        print("posição inválida ou ocupada. tente novamente")
        input("Pressione Enter para continuar...")
        continue
    except ValueError:
      print("Por favor, digite um número de 1 a 9.")
      input("Pressione Enter para continuar...")
      continue

    #Marcador de jogada
    tabuleiro[escolha] = jogador_atual
    jogadas += 1

    #Verifica se o jogador venceu
    if Vitoria(tabuleiro, jogador_atual):
      ver_tabuleiro(tabuleiro)
      print(f"Jogador {jogador_atual} venceu!")
      return

    #alterna o jogador
    jogador_atual = "X" if jogador_atual == "O" else "O"

  #empate
  ver_tabuleiro(tabuleiro)
  print("Empate!")

#inicialização do codigo para o jogo rodar
if __name__ == "__main__":
  jogar()
```

37. Funções Recursivas

Implemente uma função recursiva para calcular o fatorial de um
número.

```python
# Define a função recursiva com indicação explícita dos tipos de entrada e retorno
def fatorial(n: int) -> int:

    #exibe erro já que não existe fatorial de número negativo
    if n < 0:
        raise ValueError("Numeros negativos não possuem fatorial")

    #fatorial de 0 ou 1 é 1
    if n in (0, 1):
        return 1

    #chamada recursiva
    return n * fatorial(n - 1)

# Bloco de execução principal para testar a função
if __name__ == "__main__":
    num = 5

    # Exibe os resultados dos cálculos
    print(f"O fatorial de {num} é {fatorial(num)}")
    print(f"O fatorial de 0 é {fatorial(0)}")
```

40. Lista de Compras
o Crie um programa para gerenciar uma lista de compras, permitindo
adicionar, remover e listar itens.

```python
# Cria uma lista vazia para armazenar os itens
lista_compras = []

# Laço infinito: mantém o menu rodando até o usuário escolher sair
while True:
    # Exibe o menu principal
    print("\n--- LISTA DE COMPRAS ---")
    print("1. Adicionar item")
    print("2. Remover item")
    print("3. Visualizar lista")
    print("4. Sair")

    opcao = input("Escolha uma opção (1-4): ")

    # OPÇÃO 1: Adicionar item
    if opcao == "1":
        # .strip() remove espaços desnecessários antes e depois do texto
        item = input("Digite o nome do item para adicionar: ").strip()

        # Garante que o usuário não adicionou um texto em branco
        if item:
            lista_compras.append(item)  # Adiciona o item ao final da lista
            print(f"'{item}' foi adicionado com sucesso!")
        else:
            print("Nome do item não pode ser vazio.")

    # OPÇÃO 2: Remover item
    elif opcao == "2":
        # verifica se a lista está vazia antes de tentar remover
        if not lista_compras:
            print("A lista está vazia. Não há nada para remover.")
        else:
            print("\nItens atuais:")
            # enumerate mostra os itens numerados a partir do número 1
            for i, item in enumerate(lista_compras, 1):
                print(f"{i}. {item}")

            # try/except para privinir o programa de fechar por erro do usuario
            try:
                indice = int(input("Digite o número do item que deseja remover: "))

                # Valida se o número escolhido existe na lista
                if 1 <= indice <= len(lista_compras):
                    # .pop() remove o item e usa (indice - 1) porque listas em Python começam no 0
                    removido = lista_compras.pop(indice - 1)
                    print(f"'{removido}' foi removido com sucesso!")
                else:
                    print("Número inválido.")
            except ValueError:
                print("Por favor, digite um número válido.")

    # OPÇÃO 3: Visualizar a lista
    elif opcao == "3":
        if not lista_compras:
            print("\nSua lista de compras está vazia.")
        else:
            print("\n--- SUA LISTA ---")
            for i, item in enumerate(lista_compras, 1):
                print(f"{i}. {item}")

    # OPÇÃO 4: Sair
    elif opcao == "4":
        print("Programa encerrado. Boas compras!")
        break  # Quebra o laço 'while' e encerra a execução

    # Tratamento para opções inválidas do menu
    else:
        print("Opção inválida. Tente novamente.")
```

42. Números Primos até N

Gere todos os números primos até um número n fornecido.

```python
# Solicita um número limite ao usuário
n = int(input("Digite um número:"))

# Valida se o número digitado é maior ou igual a 2 para calcular os primos
if n >= 2:
    # Inicializa uma lista de booleanos  cobrindo os índices de 0 a n
    primo = [True] * (n + 1)

    # Marca explicitamente os índices 0 e 1 como não primos
    primo[0] = primo[1] = False

    # Percorre os números verificando apenas até a raiz quadrada de n
    for i in range(2, int(n**0.5) + 1):
        # Se o número i ainda for True, ele é primo e seus múltiplos devem ser anulados
        if primo[i]:
            # Marca os múltiplos de i como False, iniciando em i * i com incremento de i em i
            for j in range(i * i, n + 1, i):
                primo[j] = False

    # Filtra e coleta em uma nova lista todos os índices que permaneceram como True
    res = [i for i in range(2, n + 1) if primo[i]]
else:
    # Retorna uma lista vazia para valores menores que 2
    res = []

# Exibe  a lista final contendo todos os números primos encontrados
print(res)
```

47. Contar Números em Lista

Conte quantas vezes cada número aparece em uma lista.

```python
# Define a lista contendo uma coleção de números inteiros com repetições
numeros = [8, 9, 7, 4, 5, 6, 1, 2, 1, 2, 3, 6, 4, 5, 1, 2, 4, 8, 9, 8, 9, 4, 5, 6]

# Inicializa um dicionário vazio para armazenar a frequência de cada número
contagem = {}

# Cria uma nova lista contendo os elementos ordenados em ordem crescente
ordemNum = sorted(numeros)

# Percorre sequencialmente cada elemento da lista ordenada
for num in ordemNum:

  # Atualiza a contagem do número no dicionário utilizando 0 como valor padrão
    contagem[num] = contagem.get(num, 0) + 1

#exibe o dicionário final com a contagem de frequência das chaves
print(contagem)
```

48. Ordenação por Inserção

Implemente o algoritmo de ordenação por inserção (Insertion Sort).

```python
# Define a lista inicial desordenada com elementos numéricos
lista = [9, 27, 10, 5, 17, 2]

# Percorre a lista a partir do segundo elemento até o final
for i in range(1, len(lista)):
    # Armazena o valor do elemento atual que será inserido na posição correta
    key = lista[i]

    # Define o índice do elemento imediatamente anterior ao elemento chave
    j = i - 1

    # Desloca os elementos maiores que a chave para uma posição à frente
    while j >= 0 and lista[j] > key:
        lista[j + 1] = lista[j]
        j -= 1

    # Insere o elemento chave na sua posição final correta dentro da sublista ordenada
    lista[j + 1] = key

# Exibe a lista totalmente ordenada
print(lista)
```
